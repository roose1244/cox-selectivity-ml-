````python
def get_all_activities(target_id, total_records, page_size=1000):
    all_activities = []

    for offset in range(0, total_records, page_size):
        params = {
            "target_chembl_id": target_id,
            "limit": page_size,
            "offset": offset
        }

        r = requests.get(
            "https://www.ebi.ac.uk/chembl/api/data/activity.json",
            params=params
        )

        print(
            f"Retrieved {offset + 1}–"
            f"{min(offset + page_size, total_records)}"
        )

        r.raise_for_status()
        all_activities.extend(r.json()["activities"])

    return all_activities
````

````python
import time
import requests

def get_all_activities(target_id, total_records, page_size=100):
    all_activities = []

    for offset in range(0, total_records, page_size):

        params = {
            "target_chembl_id": target_id,
            "limit": page_size,
            "offset": offset
        }

        for attempt in range(3):
            try:
                r = requests.get(
                    "https://www.ebi.ac.uk/chembl/api/data/activity.json",
                    params=params,
                    timeout=30
                )

                r.raise_for_status()
                batch = r.json()["activities"]
                all_activities.extend(batch)

                print(
                    f"Retrieved {len(all_activities):,} / {total_records:,}"
                )

                break

            except requests.exceptions.RequestException:
                print(
                    f"Request failed at {offset:,}. "
                    f"Attempt {attempt + 1}/3"
                )

                if attempt < 2:
                    time.sleep(3)
                else:
                    raise

    return all_activities
             
````

# Decisions and next steps

Everything above this point is the original exploratory session: it queried the
ChEMBL API interactively, held the results in kernel memory, and wrote nothing
to disk. The counts printed above came from ChEMBL release **ChEMBL_37** and are
kept as a record of that exploration. They are *not* re-used as inputs anywhere
— the download code reads the current totals from ChEMBL's own `page_meta`.

## Endpoint decision: exact IC50 in nM

The README originally proposed **Ki** as the primary endpoint while the
exploration above filtered exact (`=`) **IC50** in nM. That discrepancy is now
resolved in favour of IC50, and the README has been updated to match.

**Decision: exact (`=`) IC50 values reported in nM.**

Why:

1. **IC50 and Ki are different quantities and must not be pooled.** Ki is an
   equilibrium dissociation constant. IC50 is a functional potency that depends
   on substrate concentration, enzyme concentration and assay readout. Mixing
   them would add spread that has nothing to do with molecular structure.
2. **IC50 is much better populated for both COX isoforms** in this ChEMBL
   release. The per-step counts are logged in
   `../data/processed/filter_log.csv`.
3. **It keeps the exploration above valid** rather than discarding it.

Ki, AC50 and percent-inhibition records are still present in the raw download
under `../data/raw/` and could be analysed separately later. They are excluded
from the curated IC50 dataset.

**Censored values** (`>`, `<`, `>=`, `<=`, `~`, or a missing relation) are
excluded. A record such as "IC50 > 100000 nM" gives no usable number for a
regression target. The count dropped at that step is logged, so the exclusion is
visible rather than silent.

## Potency and selectivity

```text
pIC50       = 9 - log10(IC50_nM)
selectivity = log10(IC50_COX-1 / IC50_COX-2) = pIC50_COX-2 - pIC50_COX-1
```

**Orientation: a positive selectivity value means stronger inhibition of COX-2
than of COX-1 (COX-2 selective).** The column is named
`selectivity_log10_IC50_cox1_over_cox2` so that the orientation is readable from
the data file itself.

Note that `-np.log10(value * 1e-9)` used in the cells above is algebraically the
same as `9 - log10(value_in_nM)`; the curation code uses the second form because
it states the unit assumption explicitly.

## Where the code now lives

The ad-hoc download functions above have been replaced by two reusable modules,
so that the pipeline can be re-run from a clean checkout:

- `../src/chembl_download.py` — verifies both target records against the live
  API, paginates through every activity record with a timeout,
  `raise_for_status`, retry with exponential backoff and progress messages, and
  writes the unmodified tables plus `provenance.json` to `../data/raw/`.
- `../src/curate.py` — applies the documented filter chain, collapses replicates
  to one median value per molecule–target pair, and builds the paired
  selectivity table in `../data/processed/`.


## Re-running the pipeline

From the project root, with the virtual environment active:

```bash
source .venv/bin/activate
pip install -r requirements.txt
python src/chembl_download.py   # writes data/raw/
python src/curate.py            # writes data/processed/
```

The download takes several minutes because ChEMBL is paginated. Both scripts are
idempotent: re-running them overwrites their own outputs and nothing else.

### Stopping point

The paired selectivity dataset is now on disk with a logged filter chain. The
next stages — RDKit descriptors and Morgan fingerprints, baseline models before
random forest or gradient boosting, a scaffold-aware split alongside a random
split, preprocessing fitted on training folds only, and SHAP — are planned but
deliberately not started here.

Two findings above should shape how those results are reported: the fraction of
pairs assembled from a single publication, and the fraction involving a molecule
with poorly reproducible replicates. Where a selectivity value is built from two
different papers, assay differences are folded into the ratio, and molecular
structure cannot account for that. Any model trained on the full paired set
should be compared against the same-publication subset, with both sample sizes
stated.
